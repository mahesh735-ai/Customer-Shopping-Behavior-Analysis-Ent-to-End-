```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x : x.fillna(x.median()) )
```

```python
# Rename columns to snake_case for consistency and safe execution across platforms like SQL.
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
# 'purchase_amount_(usd)' : This not good for future ,so make it readable
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
# creating a columns age_group

labels = ['Young Adult','Adult','Middle-aged','Senior']
df['age_group'] =pd.qcut(df['age'],q=4,labels = labels)
```

```python
df[['age','age_group']].head(10)
# Useful for creating marketing strategies and understanding customer behavior to target specific customer segments.
```

```python
df.head()
```

```python
# Create numeric column 'purchase_frequency_days' for easier analysis and matrix building.

frequency_mapping = {
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied','promo_code_used']].head(10)
# Discounts can arise from promo codes or sales; check for redundancy to ensure no double-counting.
```
